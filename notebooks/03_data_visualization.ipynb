# Interactive Market Dashboard (Streamlit)

This notebook documents the interactive dashboard built using Streamlit.
The actual application lives in:

```
    app/streamlit_app.py
```

## Why Streamlit?
- Interactive filters
- Real-time insights
- Suitable for deployment

## How to Run Locally

```
streamlit run app/streamlit_app.py
```

## Data Source
- SQLite database: data/database.db
- Table: jvc_apartments

## Features
- Bedroom / price / area filters
- KPIs
- Price distribution
- Micro-market analysis
